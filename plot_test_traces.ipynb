```python
import spikeinterface.full as si
import matplotlib.pyplot as plt
import numpy as np

# Load Open Ephys recording
recording = si.read_openephys('/Users/harryclark/Desktop/M26_D0_2024-10-24_15-14-08_1')  # <-- update path
channel_ids = recording.get_channel_ids()
recording = si.bandpass_filter(recording, freq_min=300, freq_max=6000)
recording = si.common_reference(recording, reference='global')
recording = si.depth_order(recording)
fig, ax = plt.subplots(ncols=4, nrows=1, figsize=(15, 5))
for i, ch in enumerate(channel_ids):
    lil_rec = recording.select_channels(channel_ids=[channel_ids[i]]).frame_slice(start_frame=30_000*30, end_frame=30_000*40)
    traces = lil_rec.get_traces().T
    print(f'Channel {i} traces shape: {traces.shape}')
    ax[0].plot(traces.T, i*np.ones(len(traces.T)), alpha=0.5)
plt.show()
```
